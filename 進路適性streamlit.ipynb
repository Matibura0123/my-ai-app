```python
# 💡 データファイルが存在するかを確認するセル

import os

CSV_DIR = '/content/drive/MyDrive/Colab Notebooks/職業'
required_files = ["questions_axis.csv", "profiles.csv"]

print(f"データフォルダの確認: {CSV_DIR}")
if not os.path.exists(CSV_DIR):
    print("❌ 致命的エラー: フォルダが見つかりません。Google Drive のマウントを確認してください。")
else:
    all_found = True
    for file in required_files:
        path = os.path.join(CSV_DIR, file)
        if os.path.exists(path):
            print(f"✅ {file} が見つかりました。")
        else:
            print(f"❌ {file} が見つかりません。Google Drive の配置を確認してください。")
            all_found = False

    if all_found:
        print("🎉 データファイルはすべてOKです。次のセルに進んでください。")
```
